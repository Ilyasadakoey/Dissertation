```python
import pandas as pd
import numpy as np

# ---------------------------
# Parameter
# ---------------------------
input_file = "alle_peaks_normiert.csv"      # Eingabedatei
output_file = "ausgewählte_peaks_pivot.csv" # Ausgabe
target_mz = [225, 100, 85, 43]              # gewünschte m/z-Werte
tolerance = 0.2                             # Toleranz in m/z

# ---------------------------
# CSV einlesen
# ---------------------------
df = pd.read_csv(input_file)

# Prüfen, ob Spalten vorhanden sind
required_cols = {"Messreihe", "m/z", "area_norm"}
if not required_cols.issubset(df.columns):
    raise ValueError(f"Fehlende Spalten in {input_file}: {required_cols - set(df.columns)}")

# ---------------------------
# Filterung auf Ziel-m/z
# ---------------------------
mask = np.full(len(df), False)
for mz_target in target_mz:
    mask |= np.abs(df["m/z"] - mz_target) <= tolerance

df_filtered = df[mask].copy()

# Runde m/z auf ganze Zahl, um Pivot stabil zu machen
df_filtered["m/z_rounded"] = df_filtered["m/z"].round().astype(int)

# ---------------------------
# Pivot-Tabelle erstellen
# Jede Messreihe = eine Zeile, jede m/z = Spalte mit area_norm
# ---------------------------
df_pivot = (
    df_filtered
    .pivot_table(
        index=["Messreihe"],
        columns="m/z_rounded",
        values="area_norm",
        aggfunc="mean"
    )
    .reset_index()
)

# Spaltennamen schöner machen
df_pivot.columns.name = None
df_pivot = df_pivot.rename(columns={mz: f"m/z_{mz}" for mz in df_pivot.columns if isinstance(mz, (int, float))})

# ---------------------------
# Speichern
# ---------------------------
df_pivot.to_csv(output_file, index=False)
print(f"✅ Pivotierte Tabelle gespeichert in '{output_file}'")
print(f"   Enthaltene m/z-Werte: {target_mz}")
```
